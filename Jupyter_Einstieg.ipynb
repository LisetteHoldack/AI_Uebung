## 1. Was ist ein Jupyter Notebook?

Ein Jupyter Notebook ist ein interaktives Dokument, in dem **Text, Python-Code und Ergebnisse** zusammenstehen. Es eignet sich besonders, um Daten zu untersuchen, Berechnungen auszuprobieren, Visualisierungen zu erstellen und die eigenen Schritte zu dokumentieren.

Anders als ein normales Python-Skript führst du einzelne Codezellen aus. Das Ergebnis erscheint direkt unter der jeweiligen Zelle. So kannst du eine Analyse Schritt für Schritt entwickeln und erklären.

## 2. Installation und Start

## 4. Markdown-Zellen für Text und Formeln

Markdown-Zellen enthalten Erklärungen statt ausführbarem Python. Mit `#` erstellst du Überschriften, mit `**Text**` fettgedruckten Text und mit Listen kannst du Inhalte ordnen. Mathematische Formeln schreibst du zwischen Dollarzeichen, zum Beispiel: $a^2 + b^2 = c^2$.

## 5. Beispiel: Einfache Datenanalyse mit Python

Ein Café möchte seine Tagesumsätze vergleichen. Für Tabellen und Datenanalysen verwenden viele Python-Nutzerinnen und -Nutzer die Bibliothek **pandas**. Falls pandas in deiner Python-Umgebung noch fehlt, führe die folgende `%pip install pandas`-Zelle einmal aus. `%pip` installiert das Paket in der Umgebung des aktiven Jupyter-Kernels.

Danach importieren wir pandas. `as pd` ist eine übliche Kurzform, damit wir die Bibliothek bequem ansprechen können.

````python
%pip install pandas
````

````python
import pandas as pd
````

Nun legen wir eine kleine Tabelle mit Wochentagen und Umsatz in Euro an. `DataFrame` ist pandas' Tabellenformat; `df` ist ein häufig verwendeter Variablenname dafür.

````python
umsatzdaten = {
    "Wochentag": ["Montag", "Dienstag", "Mittwoch", "Donnerstag", "Freitag"],
    "Umsatz_EUR": [180, 210, 195, 240, 325],
}

df = pd.DataFrame(umsatzdaten)
df
````

Mit `head()` schauen wir uns die ersten fünf Zeilen der Tabelle an. Bei einer größeren Tabelle kannst du zum Beispiel `df.head(3)` schreiben, um nur die ersten drei Zeilen zu sehen.

````python
df.head()
````

Zum Schluss berechnen wir mit `mean()` den durchschnittlichen Tagesumsatz. Die Ausgabe sollte `230.0` sein: Über diese fünf Tage lag der Umsatz im Mittel bei 230 Euro.

````python
durchschnitt = df["Umsatz_EUR"].mean()
print(f"Durchschnittlicher Tagesumsatz: {durchschnitt:.2f} EUR")
````
